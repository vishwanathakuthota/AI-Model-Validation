## 5. Experiment — Fitting a Numerical Relationship

The following artificial dataset relates house size to price. It is illustrative and is not a property-valuation dataset.


```python
X=np.array([600,800,1000,1200,1500,1800,2200],dtype=float)
y=np.array([42,52,63,73,88,103,124],dtype=float)

plt.figure(figsize=(8,5))
plt.scatter(X,y)
plt.xlabel("House size (sq.ft.)")
plt.ylabel("Price (₹ lakh)")
plt.title("Observed Data")
plt.grid(alpha=.25)
plt.show()
```

```python
w,b=np.polyfit(X,y,1)
def model(x): return w*x+b
print(f"Estimated slope w     = {w:.6f}")
print(f"Estimated intercept b = {b:.6f}")

xline=np.linspace(X.min(),X.max(),200)
plt.figure(figsize=(8,5))
plt.scatter(X,y,label="Observed data")
plt.plot(xline,model(xline),label="Fitted model")
plt.xlabel("House size (sq.ft.)")
plt.ylabel("Price (₹ lakh)")
plt.title("Fitted Linear Model")
plt.legend()
plt.grid(alpha=.25)
plt.show()
```

## 6. Prediction Is Not Truth

A model can produce a prediction for a new input. The computation establishes that an output was generated; it does not establish that the output equals reality.


```python
x_new=2000
prediction=model(x_new)
print(f"Input      = {x_new} sq.ft.")
print(f"Prediction = ₹{prediction:.2f} lakh")
```

Define signed prediction error as:


Absolute error is:


```python
actual=112.0
prediction=model(2000)
error=actual-prediction
print(f"Actual         = ₹{actual:.2f} lakh")
print(f"Predicted      = ₹{prediction:.2f} lakh")
print(f"Signed error   = ₹{error:.2f} lakh")
print(f"Absolute error = ₹{abs(error):.2f} lakh")
```

## 7. Mean Absolute Error

For observation i:


For n observations, Mean Absolute Error is:


MAE measures average absolute prediction error in the same units as the target variable.


```python
X_test=np.array([700,1100,1400,1700,2000,2400],dtype=float)
y_test=np.array([47,68,84,98,112,133],dtype=float)
y_pred=model(X_test)
errors=np.abs(y_test-y_pred)
mae=errors.mean()

print(" SIZE    ACTUAL   PREDICTED   ABS ERROR")
for s,a,p,e in zip(X_test,y_test,y_pred,errors):
    print(f"{int(s):>5}    {a:>6.1f}     {p:>7.2f}     {e:>7.2f}")
print(f"\nMAE = ₹{mae:.4f} lakh")
```

## 10. Data Quality Changes Model Behaviour

Parameters are estimated from data. Errors or changes in data can therefore change the fitted model.


```python
X_bad=X.copy(); y_bad=y.copy()
y_bad[-1]=300
w_bad,b_bad=np.polyfit(X_bad,y_bad,1)
def bad_model(x): return w_bad*x+b_bad

print(f"Original-data prediction  = ₹{model(2000):.2f} lakh")
print(f"Corrupted-data prediction = ₹{bad_model(2000):.2f} lakh")

xline=np.linspace(500,2300,200)
plt.figure(figsize=(8,5))
plt.scatter(X,y,label="Original observations")
plt.scatter(X_bad,y_bad,marker="x",s=90,label="Dataset with corrupted value")
plt.plot(xline,model(xline),label="Original-data model")
plt.plot(xline,bad_model(xline),label="Corrupted-data model")
plt.xlabel("House size (sq.ft.)")
plt.ylabel("Price (₹ lakh)")
plt.title("Effect of Data Quality")
plt.legend()
plt.grid(alpha=.25)
plt.show()
```
