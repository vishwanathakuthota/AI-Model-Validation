```python
import numpy as np
import matplotlib.pyplot as plt
import math
rng=np.random.default_rng(1818)
```

## 1. What Is Hallucination?

Unsupported factual claim, contradiction, fabricated citation, invented entity, incorrect attribution, stale fact, or unjustified certainty. A scientific study must define exactly which failure is counted.


## 2. Factuality

For claim c, use F(c) ∈ {Supported, Contradicted, Unresolved}. The truth standard must be defined.


## 3. Groundedness

G(c,D)=1 when supplied evidence D sufficiently supports claim c. Groundedness is relative to D; it is not universal truth.


## 4. Factuality vs Groundedness

A claim can be grounded and true, grounded in a wrong source, ungrounded but true, or ungrounded and false.


## 5. Evidence Boundary

Define whether evidence may come from supplied context, an approved knowledge base, authoritative sources, tools, model knowledge, or a stated combination.


## 6. Atomic Claims

Decompose response Y into {c1,...,cm}. Each claim should be independently verifiable where possible.


## 7. Claim Extraction

Claim extraction has its own errors: omission, duplication, over-splitting, under-splitting, and context loss.


## 8. Three Verification States

Supported, Contradicted, and Unresolved must remain distinct. Absence of evidence is not evidence of falsehood.


## 9. Entailment

Entail(D,c) asks whether evidence D sufficiently supports c.


## 10. Contradiction

Contradict(D,c) asks whether D supports an incompatible proposition. Unsupported is not the same as contradicted.


## 11. Factual Precision

Pf = supported factual claims / factual claims.


## 12. Unsupported Claim Rate

UCR = unsupported factual claims / factual claims.


## 13. Contradiction Rate

CR = contradicted factual claims / factual claims.


## 14. Factual Recall

Rf = covered required facts / required facts.


## 15. Factual F1

F1f = 2 Pf Rf / (Pf + Rf), when both precision and recall are meaningful.


## 16. Severity-Weighted Error

Ew = Σ wi·1[failure_i] / Σ wi. Weights require intended-use justification.


```python
states=np.array(["S","S","C","U","S","C"])
weights=np.array([1,2,5,2,1,10.])
failure=(states!="S").astype(float)
print("Unweighted failure:",failure.mean())
print("Severity-weighted failure:",np.sum(weights*failure)/weights.sum())
```

## 17. Citation Presence

A citation-shaped reference proves only that a citation exists. Citation presence ≠ citation support.


## 18. Citation Entailment

CE(c,d)=1 when cited source d actually entails claim c.


## 19. Citation Precision

Valid supporting citations / citations.


## 20. Citation Completeness

Cited-supported claims / factual claims requiring support.


## 21. Source Authority

Relevant source ≠ authoritative source. Evidence quality depends on the claim and domain.


## 22. Temporal Factuality

Truth(c,t) can change. Preserve claim time, source time, and evaluation time.


## 23. RAG Groundedness

Measure support of generated claims by retrieved context, while evaluating retrieval separately.


## 24. Retrieval Recall@k

Recall@k = |R∩K| / |R|.


## 25. Retrieval Precision@k

Precision@k = |R∩K| / k.


```python
relevant={"d2","d4","d7","d10"}
retrieved=["d2","d1","d7","d6","d9"]
hit=len(relevant.intersection(retrieved))
print("Recall@5:",hit/len(relevant))
print("Precision@5:",hit/len(retrieved))
```

## 26. Retrieval vs Generation Failure

Separate evidence-not-retrieved, evidence-ignored, evidence-misinterpreted, unsupported external knowledge, and correct grounded generation.


## 27. Context Utilization

Add/remove relevant evidence, distractors, and controlled contradictions; measure response changes.


## 28. Distractor Robustness

Measure changes in correctness and groundedness after adding plausible irrelevant evidence.


## 29. Conflicting Sources

Test whether the system detects conflict, attributes competing claims, reasons about authority/date, and avoids false certainty.


## 30. Knowledge Boundary

A trustworthy system distinguishes known, supported, uncertain, and unknown.


## 31. Abstention

Coverage=P(Model answers); SelectiveRisk=E[L | Model answers].


## 32. Self-Reported Confidence

Linguistic confidence is not calibrated probability. Test P(Correct | confidence≈p)≈p.


## 33. Token Probability

High token or sequence probability does not establish factual truth.


## 34. Repetition-Based Uncertainty

Disagreement can signal instability, but agreement can reproduce a shared false belief. Agreement ≠ truth.


## 35. NLI Verification

Entailment/contradiction/neutral classifiers can support verification, but the verifier itself must be validated.


## 36. LLM-as-Verifier

An LLM verifier can hallucinate, be prompt-sensitive, or misunderstand sources.


## 37. Verifier Validation

Compare automated verification with trusted adjudication using confusion matrices, precision, recall, calibration, and stability.


## 38. False Positive Rate

FPR=P(flag hallucination | claim valid).


## 39. False Negative Rate

FNR=P(mark valid | claim invalid).


## 40. Detection Threshold

Flag=1[s<τ]. Threshold choice trades false positives and false negatives.


## 41. Hallucination Denominator

Rates per response, claim, sentence, session, or task are different measurements.


## 42. Response-Level Failure

Under an illustrative independence assumption, P(any failure)=1-(1-p)^m.


```python
for p in [.005,.01,.02,.05]:
    print(p,[round(1-(1-p)**m,4) for m in [1,5,10,25,50]])
```

## 43. Length Confounding

Longer answers create more claim opportunities. Compare claim counts and claim-normalized rates.


## 44. Correlated Claim Errors

One false premise can generate many downstream errors. Multiple wrong claims are not necessarily independent failures.


## 45. Root-Cause Taxonomy

Missing knowledge, retrieval failure, conflicting context, reasoning error, entity confusion, staleness, citation fabrication, prompt ambiguity, tool failure, over-generalization.


## 46. Entity Hallucination

Test entity existence separately from attribute correctness.


## 47. Numerical Hallucination

Verify arithmetic, dates, units, percentages, counts, financial values, and statistical claims.


## 48. Executable Verification

When a claim is computationally checkable, execute the computation instead of relying on linguistic plausibility.


## 49. Citation Fabrication

Validate bibliographic existence separately from whether the source supports the claim.


## 50. Quote Verification

Verify wording, source, location, context, and meaning.


## 51. Summarization Faithfulness

Every factual summary claim should be source-supported; coverage separately measures retained important content.


## 52. Multi-Hop Factuality

Preserve chains such as d1 + d2 → c when multiple evidence pieces are required.


## 53. Causal Claims

Association ≠ causation. Unsupported causal escalation is a distinct factual failure.


## 54. Distribution Shift

Measure factuality across new domains, time periods, languages, rare entities, terminology, and long contexts.


## 55. Subgroup Factuality

Measure P(Hallucination | domain), P(Hallucination | language), and other relevant slices.


## 56. False-Premise Tests

A robust system should challenge false premises rather than manufacture supporting details.


## 57. Sycophancy

Measure whether factual accuracy drops when users confidently assert false statements.


## 58. Conflicting-Evidence Stress

Provide incompatible sources and test conflict detection, attribution, authority reasoning, and uncertainty.


## 59. Freshness

Current factuality may require retrieval/tools. Static model knowledge ≠ current fact.


## 60. Factual Risk

Risk_j = P(F_j) C_j. Equal hallucination rates can imply unequal consequences.


## 61. Expected Factual Risk

ExpectedRisk = Σ p_j C_j.


## 62. Error-Rate Uncertainty

For approximately independent claims, SE(p̂)≈sqrt[p̂(1-p̂)/n].


## 63. Cluster Bootstrap

When claims are nested in responses, resample responses rather than isolated claims.


```python
responses=300
rates=rng.beta(2,18,responses)
counts=rng.integers(3,12,responses)
errors=np.array([rng.binomial(n,p) for n,p in zip(counts,rates)])
observed=errors.sum()/counts.sum()
boot=[]
for _ in range(2000):
    idx=rng.integers(0,responses,responses)
    boot.append(errors[idx].sum()/counts[idx].sum())
print("Observed claim error:",observed)
print("Response-cluster bootstrap 95% CI:",np.quantile(boot,[.025,.975]))
```

## 64. Zero Observed Hallucinations

Under simplified independent trials, p_upper≈3/n gives an approximate 95% upper bound after zero observed failures.


## 65. Benchmark Contamination

## 66. Dynamic Factuality Evaluation

## 67. Gold Standards Are Fallible

Reference answers need provenance, review, adjudication, versioning, and correction processes.


## 68. Evidence Hierarchy

## 70. Evidence Graph
